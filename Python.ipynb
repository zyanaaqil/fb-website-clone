**5. Operators**

```python
a = 10
b = 5

print("Addition:", a + b)
print("Subtraction:", a - b)
print("Multiplication:", a * b)
print("Division:", a / b)
print("Modulus:", a % b)
```

**6. Conditional Statements (if-else)**

```python
marks = 75

if marks >= 50:
    print("Pass")
else:
    print("Fail")
```

**7. Loops**

```python
for i in range(1, 6):
    print(i)
```

**While Loop**

```python
i = 1

while i <= 5:
    print(i)
    i += 1
```

**13. Exception Handling**

```python
try:
    a = int(input("Enter first number: "))
    b = int(input("Enter second number: "))

    print(a / b)

except ZeroDivisionError:
    print("Cannot divide by zero")

except ValueError:
    print("Please enter numbers only")
```

**14. File Handling**

```python
file = open("student.txt", "w")

file.write("Welcome to Python Programming")

file.close()

file = open("student.txt", "r")

print(file.read())

file.close()
```

**17. Comparison and Logical Operators**

```python
a = 10
b = 20

print(a > b)
print(a < b)
print(a == b)

print(a < b and b > 10)
print(a > b or b > 10)
```
